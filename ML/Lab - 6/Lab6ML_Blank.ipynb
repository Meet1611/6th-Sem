# SVR

# Importing the libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
```

# Read World bank CSV

```python
df = pd.read_csv("WorldBank.csv")
```

```python
df
```

# Perform conditional selection to find - Population ages 15-64 (% of total population)

```python
data = df[df["Indicator Name"] == "Population ages 15-64 (% of total population)"]
data
```

# Divide the data into input and output

```python
x = np.arange(1960, 2020)
x = x.reshape(-1, 1)
```

```python
y = data.iloc[:,4:-1]
y = np.array(y)
y = y.reshape(-1, 1)
```

# Plot scatter plot of Population ages 15-64 (% of total population)

```python
plt.scatter(x, y)
```

# Feature Scaling (Mandatory for SVR)**
SVR is highly sensitive to the range of data points. If we don't scale (normalize) the data, the model will fail to find the correct hyperplane.

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

# Fitting SVR on Linear Kernel on dataset

```python
from sklearn.svm import SVR
```

```python
svr = SVR(kernel='linear')
svr.fit(x_scaled, y)
```

# Predict Values using Model

```python
# use inverse_transform
x_inverse = scaler.inverse_transform(x_scaled)
x_inverse
```

```python
y_pred_linear = svr.predict(x_scaled)
```

```python
plt.scatter(x_inverse, y)
plt.scatter(x_inverse, y_pred_linear)
```

# Predict the X_scaled using 3 Kernel

```python
# above Steps Only - Kernel Change
```

```python
svr = SVR(kernel="poly")
svr.fit(x_scaled, y)
y_pred_poly = svr.predict(x_scaled)
```

```python
svr = SVR(kernel="rbf")
svr.fit(x_scaled, y)
y_pred_rbf = svr.predict(x_scaled)
```

# Visualising the  results

```python
plt.figure(figsize=(10, 6))

plt.scatter(x_inverse, y, label="Data")
plt.plot(x_inverse, y_pred_linear, label="Linear")
plt.plot(x_inverse, y_pred_poly, label="Polynomial")
plt.plot(x_inverse, y_pred_rbf, label="RBF")

plt.tight_layout()
plt.xlabel("data")
plt.ylabel("target")
plt.legend()
plt.tight_layout()
plt.show()
```
